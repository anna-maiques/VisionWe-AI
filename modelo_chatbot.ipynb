##*GoodCharge*


Simulação do chatbot GoodCharge AI gerenciando os carregadores elétricos em condomínios, possibilitando que os usuários façam as perguntas sobre o consumo energético, reservas, disponibilidade, cobranças e falhas técnicas. O chat lembra das perguntas e respostas anterioras, permitindo diálogos contínuos e coerentes com o usuário.


# Perguntas e respostas esperadas
-- Perguntas

- Quanto consumi este mês?
- Tem algum carregador livre?
- Quero reservar um carregador
- Tem algum carregador sem funcionar?
- Quanto tenho que pagar este mês?

-- Respostas

- Você consumiu 120 kWh neste mês.
- O carregador 2 não tem nenhuma reserva até as 18h.
- Reserva realizada para hoje às 20h.
- O carregador 4 encontra-se offline.
-  O valor total da cobrança é R$ 154,20.

```python
!pip install openai
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get('OPENAI_API_KEY'))
```

```python
system_prompt = """
Você é uma assistente artificial especialista em carregadores de carros elétricos de condomínios, onde vc auxilia sídicos e moradores. Você responde perguntas sobre: consumo, cobrança, reservas, disponibilidade e falhas tecnicas.
Nunca invente informações, caso não tenha dados suficientes, informe ao usuário. Responda de forma curta e clara.
"""
historico = [
    {
        "role": "system",
        "content": system_prompt
    },{
        "role": "user",
            "content": "Quanto consumi este mês?"
    },{
        "role": "assistant",
         "content": "Você consumiu 120 kWh neste mês."
    },{
        "role": "user",
        "content": "Tem carregador livre?"
    },{
         "role": "assistant",
        "content": "O carregador 2 não tem nenhuma reserva até as 18h."
    },{
         "role": "user",
         "content": "Quero reservar um carregador."
     },{
          "role": "assistant",
         "content": "Reserva realizada para hoje às 20h."
    },{
         "role": "user",
         "content": "Tem algum carregador sem funcionar?"
    },{
         "role": "assistant",
        "content": "O carregador 4 encontra-se offline."
     },{
         "role": "user",
        "content": "Quanto tenho que pagar esse mês?"
     },{
         "role": "assistant",
        "content": "O valor total da cobrança é R$ 154,20."
     },{
         "role": "user",
         "content": "Existe alguma falha técnica?"
     },{
        "role": "assistant",
        "content": "O carregador 4 está offline e necessita manutenção."
     }
]

while True:
  user_pergunta = input("\nVocê:")

  if user_pergunta.lower() == "sair":
    break

  historico.append( #histórico salvo ;)
      {
          "role": "user",
          "content": user_pergunta
      }
  )
  response = client.chat.completions.create(
      model="gpt-4o-mini",
      messages=historico
  )
  resposta = response.choices[0].message.content

  historico.append(
      {
          "role": "assistant",
          "content": resposta
      }
  )

  print("\nGoodCharge AI:")
  print(resposta)
```
